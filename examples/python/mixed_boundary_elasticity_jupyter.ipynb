# Mixed-boundary linear elasticity

Run the shared installed-Python workflow, then present its common displacement output with a caller-owned figure.

```python
import eqiora.matplotlib as eqplot
from mixed_boundary_elasticity import solve
```

```python
plan, result = solve()
displacement = result.output(plan.capability.displacement)
assert result.plan_key == plan.identity
{
    "plan": plan.identity,
    "result": result.plan_key,
    "displacement_vertices": displacement.vertex_count,
}
```

```python
displacement_figure = eqplot.plot_deformed_field(
    result,
    field=plan.capability.displacement,
    scale=1.0,
)
displacement_figure
```
